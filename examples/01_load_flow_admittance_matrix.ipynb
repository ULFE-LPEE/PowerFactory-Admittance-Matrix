```python
# Connect to an existing PowerFactory project.
from src import connect
from src.adapters.powerfactory import extract_network, extract_operating_point

PROJECT_NAME = "Radial System"  # Change to your PowerFactory project name
app = connect(PROJECT_NAME, show=True)
```

```python
network = extract_network(app, base_mva=100.0, merge_closed_switches=False)
operating_point = extract_operating_point(app, network)
load_flow_y = network.get_load_flow_y_matrix(operating_point)
```

```python
import pandas as pd

pd.DataFrame(load_flow_y, index=network.bus_names, columns=network.bus_names).round(2)
```

```python
print(f"Load flow converged; {len(network.generators)} generators found.")
print(f"Passive Y-bus shape: {network.get_passive_y_matrix().shape}")
```

```python
bus_results = [
    {"Bus": name, "V (p.u.)": bus.voltage_pu, "Angle (deg)": bus.angle_deg}
    for name, bus in operating_point.buses.items()
]
pd.DataFrame(bus_results)
```
